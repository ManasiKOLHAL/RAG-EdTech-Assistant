# Abstract

Retrieval-Augmented Generation (RAG) has emerged as one of the most effective techniques for improving the factual accuracy of Large Language Models (LLMs). Instead of relying solely on the internal knowledge of an LLM, RAG systems retrieve relevant information from external knowledge bases before generating responses.

This project presents the design and implementation of an intelligent AI/ML chatbot using LangChain, ChromaDB, Groq LLM, and Sentence Transformers. The chatbot is capable of answering Artificial Intelligence and Machine Learning related questions by retrieving information from multiple heterogeneous knowledge sources including PDF documents, CSV datasets, and website URLs.

The retrieved documents are converted into embeddings using the **sentence-transformers/all-MiniLM-L6-v2** embedding model and stored inside ChromaDB. During user interaction, semantically similar documents are retrieved and supplied to the Groq-hosted **Llama 3.3 70B Versatile** model, enabling context-aware and grounded responses.

Unlike traditional chatbots that depend solely on pretrained knowledge, the proposed system ensures that answers are generated only from the indexed knowledge base. If sufficient context is unavailable, the chatbot responds with an appropriate message instead of hallucinating information.

The project demonstrates modern Retrieval-Augmented Generation architecture and provides an extensible foundation for enterprise AI assistants, educational tutors, and domain-specific question-answering systems.

# 1. INTRODUCTION

Artificial Intelligence (AI) has experienced remarkable growth over the past decade, fundamentally transforming the way humans interact with technology. From recommendation systems and virtual assistants to autonomous vehicles and intelligent healthcare systems, AI has become an integral component of modern digital solutions.

Among the most significant breakthroughs in AI is the development of **Large Language Models (LLMs)**. These models, trained on massive amounts of textual data, can understand natural language, answer complex questions, generate human-like text, summarize information, write code, and assist in numerous knowledge-intensive tasks.

Despite their impressive capabilities, standalone LLMs have several inherent limitations. They rely primarily on knowledge acquired during training and cannot dynamically access newly available information or organization-specific knowledge without retraining. Consequently, they may generate incorrect or fabricated responses, a phenomenon commonly known as **hallucination**.

To overcome these limitations, researchers introduced **Retrieval-Augmented Generation (RAG)**. RAG combines semantic search with generative AI by retrieving relevant information from external knowledge sources before generating a response. Instead of relying solely on pretrained knowledge, the model grounds its answers in retrieved documents, significantly improving factual accuracy, transparency, and explainability.

This project presents the design and implementation of an **Intelligent RAG-Based AI/ML Chatbot** using **LangChain**, **Groq LLM**, **ChromaDB**, and **Sentence Transformers**. The chatbot automatically ingests information from PDFs, CSV files, and websites, transforms them into vector embeddings, stores them in a vector database, and retrieves the most relevant content during question answering.

The proposed chatbot is designed specifically for the Artificial Intelligence and Machine Learning domain, enabling students, educators, researchers, and developers to obtain accurate, context-aware, and source-backed answers from a custom knowledge base.

The selected model provides the best balance between speed, memory efficiency, and semantic accuracy for an educational RAG application.

## PDF Loader Function

The function below:

1. Iterates through every PDF.
2. Loads all pages.
3. Converts pages into LangChain `Document` objects.
4. Adds useful metadata.
5. Logs loading progress.
6. Handles corrupted PDFs gracefully.

````python
def load_pdf_documents(pdf_directory: Path) -> List[Document]:
    """
    Load all PDF documents from a directory.

    Parameters
    ----------
    pdf_directory : Path
        Directory containing PDF files.

    Returns
    -------
    List[Document]
        List of LangChain Document objects.
    """

    documents: List[Document] = []

    pdf_files = sorted(pdf_directory.glob("*.pdf"))

    logging.info("Starting PDF loading...")

    for pdf_file in pdf_files:

        try:

            loader = PyPDFLoader(str(pdf_file))

            pages = loader.load()

            for page in pages:

                page.metadata["source"] = pdf_file.name
                page.metadata["file_path"] = str(pdf_file)
                page.metadata["document_type"] = "PDF"

                documents.append(page)

            logging.info(
                "Loaded %s (%d pages)",
                pdf_file.name,
                len(pages)
            )

        except Exception as error:

            logging.exception(
                "Failed to load %s",
                pdf_file.name
            )

            print(f"Could not load {pdf_file.name}")
            print(error)

    logging.info(
        "Total PDF pages loaded: %d",
        len(documents)
    )

    return documents
````

# CSV Loader Function

The function below:

- Reads every CSV automatically
- Converts each row into natural language
- Preserves metadata
- Handles corrupted files
- Returns LangChain Documents

Each CSV row becomes an independent retrieval unit.

````python
from typing import List
from langchain_core.documents import Document


def load_csv_documents(csv_directory: Path) -> List[Document]:
    """
    Load every CSV inside a directory and convert
    each row into a LangChain Document.

    Parameters
    ----------
    csv_directory : Path

    Returns
    -------
    List[Document]
    """

    documents: List[Document] = []

    csv_files = sorted(csv_directory.glob("*.csv"))

    logging.info("Starting CSV loading...")

    for csv_file in csv_files:

        try:

            dataframe = pd.read_csv(csv_file)

            logging.info(
                "Reading %s (%d rows)",
                csv_file.name,
                len(dataframe)
            )

            for row_number, row in dataframe.iterrows():

                row_text = "\n".join(
                    f"{column}: {value}"
                    for column, value in row.items()
                )

                document = Document(

                    page_content=row_text,

                    metadata={
                        "source": csv_file.name,
                        "file_path": str(csv_file),
                        "document_type": "CSV",
                        "row_number": int(row_number),
                        "columns": list(dataframe.columns)
                    }

                )

                documents.append(document)

        except Exception as error:

            logging.exception(
                "Failed to read %s",
                csv_file.name
            )

            print(f"Error reading {csv_file.name}")

            print(error)

    logging.info(
        "CSV Documents Loaded: %d",
        len(documents)
    )

    return documents
````

````python
# Read URLs

with open(URLS_FILE, "r", encoding="utf-8") as file:

    urls = [

        line.strip()

        for line in file

        if line.strip()

    ]

print("=" * 60)
print("URLs Found")
print("=" * 60)

for url in urls:

    print(url)

print()

print(f"Total URLs : {len(urls)}")
````

````python
# Download Website Content

website_documents = []

visited_urls = set()

headers = {

    "User-Agent":

    "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"

    " AppleWebKit/537.36 (KHTML, like Gecko)"

    " Chrome/126 Safari/537.36"

}

print("=" * 70)
print("Loading Websites")
print("=" * 70)

for url in urls:

    if url in visited_urls:
        continue

    visited_urls.add(url)

    try:

        logger.info(f"Loading {url}")

        response = requests.get(

            url,

            headers=headers,

            timeout=30

        )

        response.raise_for_status()

        soup = BeautifulSoup(

            response.text,

            "html.parser"

        )

        # Remove unnecessary tags

        for tag in soup(

            [

                "script",

                "style",

                "header",

                "footer",

                "nav",

                "noscript"

            ]

        ):

            tag.decompose()

        text = soup.get_text(

            separator=" ",

            strip=True

        )

        if len(text) < 500:

            print(f"Skipped (Too Short): {url}")

            continue

        website_documents.append(

            Document(

                page_content=text,

                metadata={

                    "source": url,

                    "type": "website"

                }

            )

        )

        print(f"Loaded : {url}")

        time.sleep(1)

    except Exception as error:

        print()

        print(f"Failed : {url}")

        print(error)

        print()
````

# 30. Text Cleaning

Raw text extracted from PDFs and websites often contains:

- Extra spaces
- Multiple newline characters
- Tabs
- HTML artifacts
- Invisible Unicode characters

These unnecessary characters reduce embedding quality and increase vector database size.

Before generating embeddings, all documents are cleaned to improve retrieval accuracy.

## Text Cleaning Function

````python
# Text Cleaning Utilities

import re
from typing import List
from langchain_core.documents import Document


def clean_text(text: str) -> str:
    """
    Clean extracted document text.

    Parameters
    ----------
    text : str
        Raw text extracted from documents.

    Returns
    -------
    str
        Cleaned text.
    """

    if not text:
        return ""

    # Remove multiple spaces
    text = re.sub(r"\s+", " ", text)

    # Remove multiple newlines
    text = re.sub(r"\n+", "\n", text)

    # Remove tabs
    text = text.replace("\t", " ")

    # Strip leading/trailing spaces
    text = text.strip()

    return text
````

# Apply Cleaning to All Documents

````python
clean_documents = []

for doc in all_documents:

    cleaned_doc = Document(
        page_content=clean_text(doc.page_content),
        metadata=doc.metadata
    )

    clean_documents.append(cleaned_doc)

print("=" * 60)
print("Cleaning Complete")
print("=" * 60)
print(f"Documents Cleaned : {len(clean_documents)}")
````

"Explain supervised learning."

Retrieved Chunk:
"Supervised learning is a machine learning technique..."

Although the wording is different, their embeddings are very similar.

The Retrieval-Augmented Generation (RAG) pipeline uses these vectors to perform semantic similarity search instead of simple keyword matching.

LangChain supports multiple retrieval strategies.

| Search Type | Description |
|-------------|-------------|
| Similarity | Returns the nearest vectors |
| MMR | Maximum Marginal Relevance (reduces redundancy) |

# Retrieval Confidence

Although ChromaDB returns similarity scores, users often prefer an easier-to-understand confidence estimate.

A simple normalization converts vector distance into an approximate confidence percentage.

This estimate should be interpreted as a retrieval confidence rather than a probability.

````python
results = reloaded_vector_store.similarity_search_with_score(

    query,

    k=5

)

confidence_scores = []

for doc, distance in results:

    confidence = max(

        0,

        min(

            100,

            (1-distance)*100

        )

    )

    confidence_scores.append(confidence)

    print("="*60)

    print(doc.metadata["source"])

    print(f"Distance : {distance:.4f}")

    print(f"Confidence : {confidence:.2f}%")
````

# 35. Prompt Engineering

Prompt engineering is one of the most important components of a Retrieval-Augmented Generation (RAG) system.

Even with an excellent retriever, a poorly designed prompt may cause the LLM to:

- Ignore retrieved documents
- Hallucinate facts
- Mix internal knowledge with retrieved knowledge
- Omit source attribution

A carefully designed system prompt ensures that responses remain grounded in the indexed knowledge base and improves the factual reliability of the chatbot.

# Prompt Design Goals

The chatbot must:

- Answer only from the retrieved context.
- Never invent information.
- Clearly state when sufficient information is unavailable.
- Mention the relevant source(s).
- Produce concise and professional explanations.
- Avoid unsupported assumptions.

# System Prompt

The following prompt is used as the system instruction for the language model.

It establishes strict rules that reduce hallucinations and encourage grounded responses.

````python
SYSTEM_PROMPT = """
You are an AI and Machine Learning assistant.

Your primary responsibility is to answer questions ONLY using the retrieved context provided to you.

Rules:

1. Never use information outside the supplied context.

2. If the answer cannot be found in the context, respond exactly:

"I don't have enough information in my knowledge base."

3. Never hallucinate.

4. Never fabricate facts, examples, or references.

5. If multiple retrieved documents provide relevant information, combine them into a coherent answer.

6. Mention the document source(s) used at the end of your answer.

7. If the retrieved context is irrelevant, politely state that sufficient information is unavailable.

8. Keep answers clear, professional, and concise.

Retrieved Context:

{context}
"""
````

# Human Prompt

The user query is inserted separately from the retrieved context.

Separating system instructions from user input improves prompt clarity and follows current prompt engineering best practices.

````python
HUMAN_PROMPT = """
Question:

{input}

Answer:
"""
````

# 37. Building the RAG Retrieval Chain

## Building the RAG Retrieval System

In this section we build the complete Retrieval-Augmented Generation (RAG) pipeline using the latest LangChain 1.x architecture.

The pipeline consists of:

- Chroma Vector Database
- Retriever
- Prompt Template
- Groq LLM
- LCEL Pipeline
- Source Retrieval

Unlike older versions of LangChain, version 1.x uses the LangChain Expression Language (LCEL), which provides a modular, efficient, and production-ready architecture.

## Retrieval-Augmented Generation (RAG)

A Retrieval-Augmented Generation (RAG) system combines two powerful components:

1. **Retriever**
   - Searches the vector database.
   - Retrieves the most relevant document chunks.

2. **Large Language Model (LLM)**
   - Reads the retrieved context.
   - Generates a grounded response.

Unlike a traditional chatbot, a RAG system answers questions using external knowledge rather than relying solely on the model's internal parameters.

This improves:

- Accuracy
- Explainability
- Source attribution
- Hallucination prevention

# RAG Architecture
## RAG Processing Pipeline

```text
                    User Question
                          │
                          ▼
                  Chroma Retriever
                          │
                Top-K Relevant Chunks
                          │
                          ▼
                Chat Prompt Template
                          │
                          ▼
          Groq Llama-3.3-70B-Versatile
                          │
                          ▼
             Grounded AI Response
                          │
                          ▼
          Answer + Sources + Confidence
```

## Build LCEL RAG Chain

````python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

prompt = ChatPromptTemplate.from_template("""
Answer only from the context.

Context:
{context}

Question:
{question}

Answer:
""")

def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

rag_chain = (
    {
        "context": retriever | format_docs,
        "question": RunnablePassthrough(),
    }
    | prompt
    | llm
    | StrOutputParser()
)
````

The Streamlit application may not launch correctly when executed directly from a Jupyter Notebook because Streamlit is designed to run as a standalone web application. Jupyter Notebook and Streamlit use different execution environments, which can lead to issues such as:

- `'streamlit' is not recognized as an internal or external command`
- Browser not opening automatically
- Port conflicts
- Environment mismatch between Jupyter Notebook and Streamlit
- Interactive interface not rendering inside the notebook

### Solution

The Streamlit application has been successfully tested using the **Anaconda Prompt** in the same Python environment used for development.

### Steps to Run the Application

1. Open **Anaconda Prompt**.
2. Activate the project environment.

# How to Chat with the AI/ML Knowledge Assistant

Follow these steps to start chatting with the chatbot:

1. Open **Anaconda Prompt**.
2. Activate the Python environment.

```bash
conda activate base
```

3. Navigate to the project folder.

```bash
cd path\to\AI_ML_RAG_CHATBOT
```

4. Launch the Streamlit application.

```bash
streamlit run app.py
```

5. Open the displayed URL (usually **http://localhost:8501**) in your web browser.

6. Type your question in the chat input box and press **Enter**.
